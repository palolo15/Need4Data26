#RAW DATA


```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv('/content/US_Accidents_March23_sampled_500k.csv')
df_raw
```

#DATA CLEANING

###DATA

```python
#Create a copy to work on cleaning data and look at data's information
df = df_raw.copy()
df.info()
df.head()
print(df.shape)
```

```python
df.duplicated().sum()
df.drop_duplicates()
```

###A. FILTER AND SET UP




```python
#Find and Calculate Missing Values as Percentage
df.isnull().sum()    #Total missing value for each column
round((df.isnull().sum() / df.shape[0]) * 100, 2)
```

```python
#Keep source 2 only
df = df[df["Source"] == "Source2"].copy()


#Fix dates
df["Start_Time"] = pd.to_datetime(df["Start_Time"].astype(str).str[:19])    #change to text(string) -> convert to time (float) / 19 characters
df["End_Time"] = pd.to_datetime(df["End_Time"].astype(str).str[:19])


#Create Target
df["High_Sev"] = [1 if severity == 3 or severity == 4 else 0     #if Severity in [3,4] -> High_Sev = 1, else 0
    for severity in df["Severity"]]



df[["Source", "Start_Time", "End_Time", "Severity", "High_Sev"]].head(20)
```

###B. Location Columns

```python
#Create Region
northeast = ["CT", "ME", "MA", "NH", "NJ",
             "NY", "PA", "RI", "VT"]

midwest = ["IL", "IN", "IA", "KS", "MI", "MN",
           "MO", "NE", "ND", "OH", "SD", "WI"]

south = ["AL", "AR", "DE", "DC", "FL", "GA",
         "KY", "LA", "MD", "MS", "NC", "OK",
         "SC", "TN", "TX", "VA", "WV"]

west = ["AK", "AZ", "CA", "CO", "HI", "ID",
        "MT", "NV", "NM", "OR", "UT", "WA", "WY"]

df["Region"] = "Unknown"

df.loc[df["State"].isin(northeast),"Region"] = "Northeast"
df.loc[df["State"].isin(midwest),"Region"] = "Midwest"
df.loc[df["State"].isin(south),"Region"] = "South"
df.loc[df["State"].isin(west),"Region"] = "West"

print(df["Region"].value_counts())     #the number of accidents at each region
print("Unknown states:")
print(df.loc[df["Region"] == "Unknown", "State"].unique())
```

```python
# Create Is_Highway
df["Is_Highway"] = 0.    #0: not a highway, 1: is a highway

highway_words = [
    "I-",
    "Interstate",
    "US-",
    "U.S.",
    "SR-",
    "State Route",
    "SH-",
    "CR-",
    "Hwy",
    "Highway",
    "Fwy",
    "Freeway",
    "Expy",
    "Expwy",
    "Expressway",
    "Pkwy",
    "Parkway",
    "Tpke",
    "Turnpike",
    "Tollway",
    "Thruway",
    "Bypass",
    "Beltway",
    "Causeway",
    "Skyway"
]

for word in highway_words:
    df.loc[df["Street"].astype(str).str.contains(
            word, case=False, na=False), "Is_Highway"] = 1

print(df["Is_Highway"].value_counts())

df["Is_Highway"]
```

```python
#Road Features:

#Convert to 1/0
road_columns = ["Junction", "Traffic_Signal", "Crossing",
                "Station", "Railway", "Amenity"]

for column in road_columns:
    df[column] = df[column].astype(int)

#Merge rarevariable
control_columns = ["Stop", "Give_Way", "Traffic_Calming", "Bump"]
df[control_columns] = df[control_columns].astype(int)
df["Traffic_Control"] = df[control_columns].sum(axis=1)

#Drop Roundabout and No_Exit
df = df.drop(columns=["Roundabout", "No_Exit"])
```

```python
#Create State_Group: State with <1000 accidents -> Other

state_counts = df["State"].value_counts()

df["State_Grp"] = df["State"]

df.loc[df["State"].map(state_counts) < 1000,
       "State_Grp"] = "Other"

print(df["State_Grp"].value_counts())
```

```python
df["Traffic_Control"].head(30)
```

###C. Time Columns

```python
#Add Year, Month, Hour, DayOfWeek, Is_Weekend
df["Year"] = df["Start_Time"].dt.year
df["Month"] = df["Start_Time"].dt.month
df["Hour"] = df["Start_Time"].dt.hour
df["DayOfWeek"] = df["Start_Time"].dt.dayofweek

df["Is_Weekend"] = 0
df.loc[df["DayOfWeek"] >= 5, "Is_Weekend"] = 1

#Add Time_of_Day
df["Time_of_Day"] = "Evening"
df.loc[df["Hour"].between(0, 5), "Time_of_Day"] = "Night"
df.loc[df["Hour"].between(6, 9), "Time_of_Day"] = "Morning rush"
df.loc[df["Hour"].between(10, 15), "Time_of_Day"] = "Midday"
df.loc[df["Hour"].between(16, 19), "Time_of_Day"] = "Evening rush"

#Is_Rush_Hour:
df["Is_Rush_Hour"] = 0

weekday = df["DayOfWeek"] < 5

rush_time = df["Time_of_Day"].isin(["Morning rush", "Evening rush"])

df.loc[weekday & rush_time, "Is_Rush_Hour"] = 1
```

###D. Weather Columns

```python
#Checking unvalid value and set to blank
df["Wind_Speed(mph)"] = pd.to_numeric(df["Wind_Speed(mph)"],errors="coerce") #change Wind_Speed datatype to float


valid_ranges = {"Temperature(F)": (-40, 130),
                "Humidity(%)": (0, 100),
                "Pressure(in)": (20, 32),
                "Visibility(mi)": (0, 50),
                "Wind_Speed(mph)": (0, 100),
                "Precipitation(in)": (0, 5)}

for column, (minimum, maximum) in valid_ranges.items():
    df.loc[~df[column].between(minimum, maximum),column] = np.nan
```

```python
#Fill precipitation (Missing 47% from the missing table above)
weather = df["Weather_Condition"].astype(str).str.lower()

missing = df["Precipitation(in)"].isna()
no_precipitation = ~weather.str.contains("rain|snow|storm", na=False)

# No rain, snow, or storm -> 0
df.loc[missing & no_precipitation, "Precipitation(in)"] = 0

# The others = median
df["Precipitation(in)"] = df["Precipitation(in)"].fillna(
    df["Precipitation(in)"].median())

df["Precipitation(in)"].head(30)
```

```python
#Sunrise_Sunset from hour

# Find the missing or impossible value
missing_sun = ~df["Sunrise_Sunset"].isin(["Day", "Night"])

#Fill by hour
df.loc[missing_sun, "Sunrise_Sunset"] = "Night"
df.loc[missing_sun & df["Hour"].between(6, 19), "Sunrise_Sunset"] = "Day"

# Create Is_Night
df["Is_Night"] = (df["Sunrise_Sunset"] == "Night").astype(int)
```

```python
#Group Weather_Condition:
#Weather_Condition already string

condition = (df["Weather_Condition"].astype(str).str.lower())



df["Weather_Group"] = "Unknown"

df.loc[condition.str.contains("clear|fair", na=False), "Weather_Group"] = "Clear/Fair"
df.loc[condition.str.contains("cloud|overcast", na=False), "Weather_Group"] = "Cloudy"
df.loc[condition.str.contains("rain|drizzle|shower", na=False), "Weather_Group"] = "Rain"
df.loc[condition.str.contains("fog|mist|haze", na=False), "Weather_Group"] = "Fog/Low vis"
df.loc[condition.str.contains("snow|ice|sleet|hail", na=False), "Weather_Group"] = "Snow/Ice"
df.loc[condition.str.contains("thunder|storm", na=False), "Weather_Group"] = "Thunderstorm"

#Bad_Weather:
bad_weather = ["Rain", "Snow/Ice", "Thunderstorm", "Fog/Low vis"]
df["Bad_Weather"] = (df["Weather_Group"].isin(bad_weather)).astype(int)   #1/0 -> 1: rain, snow/ice, thunderstorm, fog/low vis
print(df["Weather_Group"].value_counts())

df["Weather_Group"]
```

```python
#Clean Wind_Direction:

wind = df["Wind_Direction"].str.upper()

directions = {"N": "North",
              "S": "South",
              "E": "East",
              "W": "West"}

df["Wind_Direction_Clean"] = (wind.str[0].map(directions).fillna("Unknown"))

df.loc[wind.isin(["CALM", "C"]), "Wind_Direction_Clean"] = "Calm"

df.loc[wind.isin(["VAR", "VARIABLE"]), "Wind_Direction_Clean"] = "Variable"

print(df["Wind_Direction_Clean"].value_counts())

df["Wind_Direction_Clean"]
```

```python
#Flag thin combinations(Regions and Weather):
group_cols = ["Region", "Weather_Group"]

cell_size = df.groupby(group_cols)["ID"].transform("count")

df["Thin_Cell"] = (cell_size < 500).astype(int)

print(df["Thin_Cell"].value_counts())

#Combination Table
combination_table = (df.groupby(group_cols).agg(Accident_Count=("ID", "count"),
                                                Thin_Cell=("Thin_Cell", "first")).reset_index())
combination_table
```

###E. Remove and save

```python
#Drop columns
columns_to_drop = ["End_Lat", "End_Lng", "Country", "Turning_Loop", "Wind_Chill(F)",
                   "Civil_Twilight", "Nautical_Twilight", "Astronomical_Twilight",
                   "Airport_Code", "Weather_Timestamp"]

df = df.drop(columns=columns_to_drop, errors="ignore")

df
```

```python
#Save accidents_clean.cvs
df.to_csv("accidents_clean.csv", index=False)
print("accidents_clean.csv has been saved!")
```

```python
#Save model_ready.cvs
model_columns = ["Temperature(F)", "Humidity(%)", "Pressure(in)",
                 "Visibility(mi)", "Wind_Speed(mph)", "Precipitation(in)",
                 "Precip_Missing", "Is_Highway", "Junction", "Traffic_Signal",
                 "Crossing", "Station", "Railway", "Amenity", "Traffic_Control",
                 "Year", "Month", "Hour", "DayOfWeek", "Is_Weekend",
                 "Is_Rush_Hour", "Is_Night", "Bad_Weather", "Thin_Cell",
                 "Region", "State_Grp", "Time_of_Day", "Weather_Group",
                 "Wind_Direction_Clean"]

category_columns = ["Region", "State_Grp", "Time_of_Day",
                    "Weather_Group", "Wind_Direction_Clean"]

model_df = df[model_columns + ["High_Sev"]].copy()

model_df = pd.get_dummies(model_df, columns=category_columns, dtype=int).fillna(0)

print("Missing values:", model_df.isna().sum().sum())

model_df.to_csv("model_ready.csv", index=False)

print("model_ready.csv has been saved!")
```
